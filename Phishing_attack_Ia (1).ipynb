```python

#installing the  required pacakages
!pip install xgboost -q

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_curve,
    auc
)

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from xgboost import XGBClassifier


# 1. DATASET UPLOADIG  AND READING

uploaded_file = files.upload()

csv_name = list(uploaded_file.keys())[0]
phish_data = pd.read_csv(csv_name)

print("Dataset loaded into Colab successfully.")
print("Rows and columns:", phish_data.shape)
display(phish_data.head())

# 2. INITIAL DATA UNDERSTANDING


print("\nAvailable columns:")
print(phish_data.columns.tolist())

print("\nMissing value count:")
print(phish_data.isnull().sum())

print("\nDataset summary:")
display(phish_data.describe())


# 3. DATA PREPARATION


# Remove ID column if it is present
if "Index" in phish_data.columns:
    phish_data = phish_data.drop(columns=["Index"])

label_name = "class"

if label_name not in phish_data.columns:
    raise Exception("The dataset must contain a target column named 'class'.")

input_features = phish_data.drop(columns=[label_name])
output_label = phish_data[label_name]

# Converting the class values:
# -1 = phishing website
#  1 = legitimate website
output_label = output_label.replace({-1: 0, 1: 1})

print("\nClass count after conversion:")
print(output_label.value_counts())


# 4. TRAINING AND TESTING SPLIT

train_x, test_x, train_y, test_y = train_test_split(
    input_features,
    output_label,
    test_size=0.20,
    random_state=25,
    stratify=output_label
)

print("\nTraining samples:", train_x.shape[0])
print("Testing samples:", test_x.shape[0])



# 5. NORMALIZATION FOR SCALE-SENSITIVE MODELS

normalizer = StandardScaler()

train_x_norm = normalizer.fit_transform(train_x)
test_x_norm = normalizer.transform(test_x)


# 6. CLASSIFICATION MODELS

classifier_set = {
    "Logistic Regression": LogisticRegression(max_iter=1200),
    "Naive Bayes": GaussianNB(),
    "Support Vector Machine": SVC(probability=True),
    "K-Nearest Neighbors": KNeighborsClassifier(n_neighbors=5),
    "Decision Tree": DecisionTreeClassifier(random_state=25),
    "Random Forest": RandomForestClassifier(
        n_estimators=250,
        random_state=25
    ),
    "XGBoost": XGBClassifier(
        eval_metric="logloss",
        random_state=25
    )
}

# 7. MODEL TRAINING, TESTING, AND EVALUATION

evaluation_records = []
saved_classifiers = {}

scale_based_models = [
    "Logistic Regression",
    "Naive Bayes",
    "Support Vector Machine",
    "K-Nearest Neighbors"
]

for clf_name, clf_model in classifier_set.items():

    print("\n" + "-" * 70)
    print("Evaluating:", clf_name)
    print("-" * 70)

    if clf_name in scale_based_models:
        clf_model.fit(train_x_norm, train_y)
        predicted_y = clf_model.predict(test_x_norm)
        probability_y = clf_model.predict_proba(test_x_norm)[:, 1]
        fold_scores = cross_val_score(clf_model, train_x_norm, train_y, cv=5)
    else:
        clf_model.fit(train_x, train_y)
        predicted_y = clf_model.predict(test_x)
        probability_y = clf_model.predict_proba(test_x)[:, 1]
        fold_scores = cross_val_score(clf_model, train_x, train_y, cv=5)

    saved_classifiers[clf_name] = clf_model

    acc_value = accuracy_score(test_y, predicted_y)
    prec_value = precision_score(test_y, predicted_y)
    rec_value = recall_score(test_y, predicted_y)
    f1_value = f1_score(test_y, predicted_y)
    cv_value = fold_scores.mean()

    evaluation_records.append({
        "Algorithm": clf_name,
        "Accuracy (%)": acc_value * 100,
        "Precision (%)": prec_value * 100,
        "Recall (%)": rec_value * 100,
        "F1 Score (%)": f1_value * 100,
        "CV Accuracy (%)": cv_value * 100
    })

    print("Accuracy :", round(acc_value * 100, 2), "%")
    print("Precision:", round(prec_value * 100, 2), "%")
    print("Recall   :", round(rec_value * 100, 2), "%")
    print("F1 Score :", round(f1_value * 100, 2), "%")
    print("CV Score :", round(cv_value * 100, 2), "%")

    print("\nDetailed Classification Report:")
    print(classification_report(test_y, predicted_y))

    matrix_output = confusion_matrix(test_y, predicted_y)

    matrix_chart = ConfusionMatrixDisplay(
        confusion_matrix=matrix_output,
        display_labels=["Phishing", "Legitimate"]
    )

    matrix_chart.plot()
    plt.title("Confusion Matrix for " + clf_name)
    plt.show()

# 8. FINAL RESULT TABLE

final_scores = pd.DataFrame(evaluation_records)
final_scores = final_scores.sort_values(
    by="Accuracy (%)",
    ascending=False
)

print("\nFinal Performance Comparison:")
display(final_scores)


# 9. ACCURACY VISUALIZATION

plt.figure(figsize=(11, 6))
plt.bar(final_scores["Algorithm"], final_scores["Accuracy (%)"])
plt.xlabel("Algorithms")
plt.ylabel("Accuracy Percentage")
plt.title("Accuracy Comparison of Phishing Detection Models")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 10. PRECISION, RECALL, AND F1 SCORE GRAPH

score_chart_data = final_scores.set_index("Algorithm")[
    ["Precision (%)", "Recall (%)", "F1 Score (%)"]
]

score_chart_data.plot(kind="bar", figsize=(12, 6))
plt.title("Precision, Recall, and F1 Score Comparison")
plt.xlabel("Algorithms")
plt.ylabel("Score Percentage")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 11. ROC CURVE ANALYSIS

plt.figure(figsize=(10, 7))

for clf_name, clf_model in saved_classifiers.items():

    if clf_name in scale_based_models:
        probability_output = clf_model.predict_proba(test_x_norm)[:, 1]
    else:
        probability_output = clf_model.predict_proba(test_x)[:, 1]

    false_rate, true_rate, cutoffs = roc_curve(test_y, probability_output)
    area_score = auc(false_rate, true_rate)

    plt.plot(
        false_rate,
        true_rate,
        label=f"{clf_name} AUC = {area_score:.3f}"
    )

plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison for Phishing Detection")
plt.legend()
plt.grid(True)
plt.show()

# 12. BEST MODEL SUMMARY


top_result = final_scores.iloc[0]

print("\nBest Model Based on Accuracy:")
print("Model Name :", top_result["Algorithm"])
print("Accuracy   :", round(top_result["Accuracy (%)"], 2), "%")
print("Precision  :", round(top_result["Precision (%)"], 2), "%")
print("Recall     :", round(top_result["Recall (%)"], 2), "%")
print("F1 Score   :", round(top_result["F1 Score (%)"], 2), "%")


# 13. FEATURE IMPORTANCE USING RANDOM FOREST

forest_classifier = saved_classifiers["Random Forest"]

importance_table = pd.DataFrame({
    "Feature Name": input_features.columns,
    "Importance Value": forest_classifier.feature_importances_
})

importance_table = importance_table.sort_values(
    by="Importance Value",
    ascending=False
)

print("\nTop 10 Important Features:")
display(importance_table.head(10))

plt.figure(figsize=(10, 6))
plt.barh(
    importance_table.head(10)["Feature Name"],
    importance_table.head(10)["Importance Value"]
)
plt.xlabel("Importance Value")
plt.ylabel("Feature Name")
plt.title("Top 10 Feature Importance - Random Forest")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

```
